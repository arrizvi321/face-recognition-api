```python
from deepface import DeepFace
import os
```

```python
img1path = os.path.join("registered", "me", "me1.jpg")
img2path = os.path.join("registered", "me", "me2.jpg")
```

```python
DeepFace.verify(img1path, img2path, model_name="SFace")
```

```python
v1 = DeepFace.represent(img1path)
v2 = DeepFace.represent(img2path)
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from PIL import Image

im = Image.open(img1path)
fig, ax = plt.subplots()

# display the image
ax.imshow(im)

# draw a rectangle around the face
face_coord = v1[0]['facial_area']
rect = patches.Rectangle((face_coord['x'], face_coord['y']), 
                         face_coord['w'], face_coord['h'], 
                         linewidth = 2, 
                         edgecolor = 'r', 
                         facecolor = 'none')

# add the patch to the axes
ax.add_patch(rect)
```

```python
DeepFace.analyze(img_path = img1path, 
                 actions = ["age", "gender", "emotion", "race"])
```
